## Data Structures and Functions

#### Work with data structures and write functions for data transformations (e.g., sum of squares, filtering).

#### Sum Of square(using List)

```python
def sum_of_squares(numbers):
    total = 0
    for n in numbers:
        total += n ** 2
    return total

nums = [1, 2, 3, 4, 5]
print(sum_of_squares(nums))
```

#### Filtering Data(Only even Numbers)

```python
def filter_even(numbers):
    return [n for n in numbers if n % 2 == 0]

nums = [1, 2, 3, 4, 5, 6]
print(filter_even(nums))
```

#### Filtering + Sum of Squares

```python
def sum_of_squares_filtered(numbers, threshold):
    return sum(n ** 2 for n in numbers if n > threshold)

nums = [1, 2, 3, 4, 5]
print(sum_of_squares_filtered(nums, 3))
```

#### Dictionary (Data Transformation)

```python
scores = {"A": 80,"B": 45,"C": 60,"D": 30,"E":75}

# Filter students with score >= 50
passed = {k: v for k, v in scores.items() if v >= 50}
print(passed)
```

#### Set (Remove Duplicates + Transform)

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

unique_squares = {n ** 2 for n in numbers}
print(unique_squares)
```

#### Sum of Square using Lambda Function

```python
numbers = [1, 2, 3, 4, 5]

sum_of_squares = sum(map(lambda x: x**2, numbers))
print(sum_of_squares)
```

#### Filtering using lambda (Even Numbers)

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
```

#### Sum Of square using Lambda & filter

```python
numbers = [1, 2, 3, 4, 5]

result = sum(
    map(lambda x: x**2,
        filter(lambda x: x > 3, numbers))
)

print(result)
```

```python
result = sum(map(lambda x: x**2, filter(lambda x: x > 3, [1, 2, 3, 4, 5])))
print(result)
```

## Client Project
#### Write a script for data cleaning (e.g., remove duplicates, filter data).

```python
raw_data = [
    {"id": 1, "name": "Amit", "age": 25, "salary": 40000},
    {"id": 2, "name": "Neha", "age": 30, "salary": 50000},
    {"id": 1, "name": "Amit", "age": 25, "salary": 40000},  
    {"id": 3, "name": "Ravi", "age": -1, "salary": 35000}, 
    {"id": 4, "name": "Pooja", "age": 28, "salary": 0}    
]

# Step 1: Remove duplicates using ID
unique_data = {row["id"]: row for row in raw_data}.values()

# Step 2: Filter valid records
clean_data = list(
    filter(lambda x: x["age"] > 0 and x["salary"] > 0, unique_data)
)

# Step 3: Final cleaned data
for record in clean_data:
    print(record)
```
