## Data Visualization with Matplotlib and Seaborn
#### Create visualizations for dataset analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample dataset
data = {
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Salary": [40000, 50000, 45000, 60000, 48000, 62000],
    "Age": [25, 30, 28, 35, 32, 40]
}

df = pd.DataFrame(data)

# 1️ Bar Chart – Average Salary by Department

avg_salary = df.groupby("Department")["Salary"].mean()

avg_salary.plot(kind="bar", title="Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()

# 2️ Line Chart – Age Trend

df["Age"].plot(kind="line", title="Employee Age Trend")
plt.xlabel("Employee Index")
plt.ylabel("Age")
plt.show()


# 3️ Histogram – Salary Distribution

df["Salary"].plot(kind="hist", bins=5, title="Salary Distribution")
plt.xlabel("Salary")
plt.show()


# 4️ Scatter Plot – Age vs Salary

plt.scatter(df["Age"], df["Salary"])
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

## Client Project: 
#### Create a dashboard for visualizing relationships between features in a dataset (e.g., scatter plots, histograms)

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create Dataset

data = {
    "Age": [22, 25, 28, 30, 35, 40, 45, 50],
    "Salary": [30000, 40000, 45000, 50000, 60000, 70000, 80000, 90000],
    "Experience": [1, 3, 4, 6, 8, 12, 15, 20]
}

df = pd.DataFrame(data)

# 1️ Scatter Plot – Age vs Salary

plt.figure()
plt.scatter(df["Age"], df["Salary"])
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()

# 2️ Scatter Plot – Experience vs Salary

plt.figure()
plt.scatter(df["Experience"], df["Salary"])
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()

# 3️ Histogram – Salary Distribution

plt.figure()
plt.hist(df["Salary"], bins=5)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()

# 4️ Histogram – Age Distribution

plt.figure()
plt.hist(df["Age"], bins=5)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```
