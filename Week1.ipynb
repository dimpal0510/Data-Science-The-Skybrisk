## Introduction to Python Programming

#### Write basic Python programs (e.g., temperature converter, calculator).

```python
celsius = int(input("Enter the Temperature in Deghrees that you want to convert: "))
feranheit = (celsius * 9/5)+32

print("The converted temperature is: ",feranheit)
```

## Simple Calculator

```python
print("1 - Add")
print("2 - subtract")
print("3 - Multiply")
print("4 - Divide")
option = int(input("Choose an operation"))
result = 0
if(option in [1,2,3,4]):
    num1 = int(input("Enter the first Number: "))
    num2 = int(input("Enter the second number: "))

    if(option == 1):
        result = num1 + num2
    elif(option == 2):
        result = num1 - num2
    elif(option == 3):
        result = num1 * num2
    elif(option == 4):
        result = num1 // num2
else:
    print("Invalid Operation entered")

print("The result of the operation is {}".format(result))
```

## Client Program
#### Create a basic data processing script (e.g., calculating the average temperature)

```python
mon = float(input("Enter Temperature of Monday"))
tue = float(input("Enter Temperature of Tuesday"))
wed =float(input("Enter Temperature of Wednesday"))
thu = float(input("Enter Temperature of Thursday"))
fri = float(input("Enter Temperature of Friday"))
sat = float(input("Enter Temperature of Saturday"))
sun = float(input("Enter Temperature of sunday"))

avg = (mon + tue + wed + thu + fri + sat + sun)/7

print("Average Temperature of Week: ",avg)
```
