## NumPy and Pandas for Data Manipulation

#### Perform operations with NumPy and manipulate datasets with Pandas.

```python
import numpy as np
import pandas as pd


# NumPy 

numbers = np.array([10, 20, 30, 40, 50])

# Basic calculations
sum_val = np.sum(numbers)
mean_val = np.mean(numbers)
square_vals = numbers ** 2

print("NumPy Sum:", sum_val)
print("NumPy Mean:", mean_val)
print("NumPy Squares:", square_vals)

# Filtering with NumPy
filtered_nums = numbers[numbers > 25]
print("Filtered NumPy Data:", filtered_nums)

# Pandas Data Manipulation

data = {
    "Employee": ["Amit", "Neha", "Ravi", "Pooja", "Amit"],
    "Age": [25, 30, 28, 35, 25],
    "Salary": [40000, 50000, 45000, 60000, 40000]
}

df = pd.DataFrame(data)

print("\nOriginal DataFrame:")
print(df)

# Remove duplicates
df = df.drop_duplicates()

# Filter data (Salary > 45000)
filtered_df = df[df["Salary"] > 45000]

# Add new calculated column
df["Salary_Squared"] = df["Salary"] ** 2

print("\nCleaned DataFrame:")
print(df)

print("\nFiltered DataFrame:")
print(filtered_df)
```

## Client Project: 
#### Clean and aggregate a dataset (e.g., remove missing values,calculate averages).

```python
import pandas as pd

# Sample dataset with missing values
data = {
    "Employee": ["Amit", "Neha", "Ravi", "Pooja", "Karan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Age": [25, 30, None, 35, None],
    "Salary": [40000, 50000, 45000, None, 48000]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Data Cleaning

# Remove rows with missing values
clean_df = df.dropna()

print("\nCleaned Data (No Missing Values):")
print(clean_df)


# Aggregation

# Calculate average age and salary by department
avg_df = clean_df.groupby("Department").agg(
    Avg_Age=("Age", "mean"),
    Avg_Salary=("Salary", "mean")
)

print("\nAggregated Data (Averages):")
print(avg_df)
```
